```python
import pandas as pd

df = pd.read_csv('../data/CT_GTFS/stop_times.txt') # reading stop times
#print(df.columns.tolist())
#print(df.head())
#print(df.shape)
#print(df['trip_id'].nunique())

routes = pd.read_csv('../data/CT_GTFS/routes.txt') # reading info on transit stops
#print(routes.head())

trips = pd.read_csv('../data/CT_GTFS/trips.txt') # reading dataset of trips

merged = df.merge(trips[['trip_id','route_id']], on='trip_id')
merged = merged.merge(routes[['route_id','route_short_name']], on='route_id') # merge these columns in one dataframe

#print(merged[['route_short_name', 'arrival_time', 'stop_id', 'stop_sequence']].head(10))
print(merged['route_short_name'].value_counts()) # counts the number of stops
hour = (merged['arrival_time'].str[:2])
busy_hour = hour.value_counts()
print(busy_hour.head(5)) # prints the top 5 busiest hours
```
